## File Operation
File handling is a crucial part of any programming language. Python provides built in functions and methods to read from and write to files, both text and binary.

```python
## Read a whole file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip()) ## strip() is used to remove the newline character at the end of each line
```

```python
## writing a file
with open('example.txt', 'w') as file:
    file.write('hello Mark\n')
    file.write('welcome to python programming\n')
```

```python
## Write a file without overriding

with open('example.txt', 'a') as file:
    file.write('This is a new data being appended to the file\n')
    file.write('Welcome to file handling\n')
```

```python
## writing list of lines to a file
lines  = ['Lin1\n', 'Line2\n', 'Line3\n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

```python
### BinarY Files
data = b'\x00\x01\x02\x03\x04'
with open('example.bin','wb') as file:
    file.write(data)
```

```python
## Read the content from a source text file and write it to a new text file

with open('example.txt', 'r') as file:
    content = file.read()

with open('new_example.txt', 'w') as file:
    file.write(content)
```

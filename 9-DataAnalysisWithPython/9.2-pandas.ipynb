### Pandas
Pandas is a powerful data manipulation library in Python, mainly used for data analysis and data cleaning. It provides two primary data primary data structures: Series and data frame.
A series is a 1D array like object, while DataFrame is 2D array, size mutable and potentially heterogeneous tabular data structure with labeled axes.

```python
import pandas as pd
```

```python
## Series - 1D labeled array
data = [1, 2, 3, 4, 5]
series = pd.Series(data)
print('series:\n', series)
print(type(series))
```

```python
## Create a series from a dictionary
data = {'a' : 1, 'b' : 2, 'c' : 3}
series = pd.Series(data)
print('series from dictionary:\n', series)
```

```python
data = [10, 45, 78, 98]
index = ['a', 'b', 'c', 'd']
series = pd.Series(data, index = index)
print('series with custom index:\n', series)
```

```python
## DataFrame - 2D labeled data structure
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print('DataFrame:\n', df)
print(type(df))
```

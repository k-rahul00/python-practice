## Generators

Generators are a simpler way to create iterators. They use the yield keyword to produce a series of values lazily which means they generate the values on the fly and do not store them in the memory

```python
def square(n):
    for i in range(3):
        return  i ** 2

square(3)
print(square(3))
```

```python
def square(n):
    for i in range(3):
        yield  i ** 2

square(3)
print(list(square(3)))
```
